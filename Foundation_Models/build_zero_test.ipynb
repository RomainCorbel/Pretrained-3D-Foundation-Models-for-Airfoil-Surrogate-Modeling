```python
import os, os.path as osp
import pandas as pd
import numpy as np

# -------- paths --------
in_csv = osp.join(os.getcwd(), "extracted_features", "cls_model_35_centered", "cls_model_35_centered_features.csv")
out_dir = osp.join(os.getcwd(), "extracted_features", "zero_features_cls")
os.makedirs(out_dir, exist_ok=True)

out_csv  = osp.join(out_dir, "zero_features_cls.csv")
out_parq = osp.join(out_dir, "zero_features_cls.parquet")
# -----------------------

# Load existing features
df = pd.read_csv(in_csv)

# Identify feature columns (everything except metadata)
meta_cols = ["foil_name", "label"]
feat_cols = [c for c in df.columns if c not in meta_cols]

# Zero out features
df_zero = df.copy()
df_zero[feat_cols] = 0.0

# Save
df_zero.to_csv(out_csv, index=False)
try:
    df_zero.to_parquet(out_parq, index=False)
except Exception as e:
    print(f"Parquet skipped ({e})")

print(f"[ZERO] Saved CSV: {out_csv}")
print(f"[ZERO] Saved Parquet: {out_parq}")
print(f"[ZERO] Shape: {df_zero.shape}")
print(f"[ZERO] #features: {len(feat_cols)}")
```
