```python
# === ShapeNet Domain Statistics – Single-Cell EDA (NO SAVING) ===

import os, os.path as osp
import numpy as np
import pandas as pd
import torch
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt
from sklearn.neighbors import NearestNeighbors

from UTILS.shapenet_dataset import ShapenetDataset

def show_boxplot(class_choice,data_folder):
    ROOT = os.getcwd()
    npoints = 50
    batch_size = 16

    # =========================
    # LOAD DATASET
    # =========================
    dataset_root = osp.join(str(ROOT), data_folder)

    ds = ShapenetDataset(
        dataset_root,
        class_choice=class_choice,
        npoints=npoints,
        split="test",
        classification=True,
        normalize=False
    )

    dl = DataLoader(ds, batch_size=batch_size, shuffle=False)

    print("Dataset size:", len(ds))

    # =========================
    # COLLECT ALL POINTS
    # =========================
    all_points = []
    all_labels = []
    all_names = []

    for batch in dl:
        pts = batch[0].numpy()
        lbl = batch[1].numpy()
        all_points.append(pts)
        all_labels.append(lbl)

    all_points = np.concatenate(all_points, axis=0)
    all_labels = np.concatenate(all_labels, axis=0)

    all_names = [osp.splitext(osp.basename(seg))[0] for *_, seg, _ in ds.datapath]


    print("All points shape:", all_points.shape)

    # =========================
    # GLOBAL COORDINATE STATS
    # =========================
    XYZ = all_points.reshape(-1, 3)

    plt.figure()
    plt.boxplot([XYZ[:,0], XYZ[:,1], XYZ[:,2]], labels=["X", "Y", "Z"])
    plt.title(f"Global X / Y / Z Boxplot  {class_choice}")
    plt.show()
```

```python
class_choices = ["default","centered","normalized","torus_default","torus_normalized", "Extruded_default_thin_uniform","Extruded_normalized_thin_uniform","OT_2D","OT_3D"]

for class_choice in class_choices:
    show_boxplot(class_choice = class_choice,data_folder = "shapenet_like_out3")
```
