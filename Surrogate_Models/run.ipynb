```python
from main import *
import os
import traceback
```

```python
baselines_to_run = [
    "GraphSAGE",
    "GUNet",
    "PointNet",
    "MLP",
]

models_to_run_global = [
    "GraphSAGEGlobal",
    "GUNetGlobal",
    "PointNetGlobal",
    "MLPGlobal",
]

datasets_pointnet = [
    "cls_model_35_centered",
    "cls_model_35_default",
    "cls_model_35_normalized",
    "cls_model_35_Extruded_default_thin_uniform",
    "cls_model_35_OT_3D",
    "cls_model_35_torus_default"
]

datasets_pointmae = [
    "PointMAE_pretrain_centered",
    "PointMAE_pretrain_default",
    "PointMAE_pretrain_normalized",
    "PointMAE_pretrain_Extruded_normalized_thin_uniform",
    "PointMAE_pretrain_OT_3D",
    "PointMAE_pretrain_torus_normalized"
]
```

```python
for model in baselines_to_run:
    run_training_pipeline1(
        model=model,
        task="scarce",
        dataset_save_dir="save_dataset/baseline_scarce",
        scores_dir="baseline_scarce/scores",
        metrics_dir="baseline_scarce/metrics",
        param="params_pointnet", # not important for baselines as they are the same in both yaml
    )
```

```python
for dataset in datasets_pointmae:
    for model in models_to_run_global:
        dataset_save_dir = f"save_dataset/{dataset}"
        scores_dir_1  = f"Trained_models/scores/{dataset}"
        metrics_dir_1 = f"Trained_models/metrics/{dataset}"

        os.makedirs(scores_dir_1, exist_ok=True)
        os.makedirs(metrics_dir_1, exist_ok=True)
        
        for run_id in range(1):
            try:
                print(f"Running | {model} | run {run_id + 1}/5")
                run_training_pipeline2(
                    model=model,
                    task="scarce",
                    dataset_save_dir=dataset_save_dir,
                    scores_dir=scores_dir_1,
                    metrics_dir=metrics_dir_1,
                    param="params_pointmae",
                    features_dir=f"../Foundation_Models/extracted_features/{dataset}/{dataset}_features.parquet",
                )
            except Exception as e:
                    print(f"❌ ERROR | {model} | {dataset} | run {run_id + 1}")
                    print(e)
                    traceback.print_exc()
                    continue
```
